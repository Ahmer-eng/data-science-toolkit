```python
import numpy as np

# 1. Create a numpy array of size 2 x 2
array_2x2 = np.array([[1, 2], [3, 4]])
print("Task 1 - 2x2 Array:\n", array_2x2)
```

```python
import numpy as np

# 2. Create a random numpy array of size 3 x 4
random_3x4 = np.random.rand(3, 4)
print("\nTask 2 - Random 3x4 Array:\n", random_3x4)
```

```python
# 3. Create a numpy array of ones and a numpy array of zeroes
ones_array = np.ones((2, 3))
zeros_array = np.zeros((2, 3))

print("\nTask 3 - Ones Array:\n", ones_array)
print("Zeros Array:\n", zeros_array)
```

```python
# 4. Create a random numpy array of size 6x3 and transpose it
random_6x3 = np.random.rand(6, 3)
transposed_array = random_6x3.T  # You can also use np.transpose(random_6x3)

print("\nTask 4 - Original Shape:", random_6x3.shape)
print("Transposed Shape:", transposed_array.shape)
print("Transposed Array:\n", transposed_array)
```

```python
# 5. Create two random numpy arrays of size n x p and p x m and compute dot product
n, p, m = 3, 4, 2
array_nxp = np.random.rand(n, p)
array_pxm = np.random.rand(p, m)

# Compute dot product using np.dot() or the @ operator
dot_product = np.dot(array_nxp, array_pxm)

print("\nTask 5 - Array n x p shape:", array_nxp.shape)
print("Array p x m shape:", array_pxm.shape)
print("Dot Product Result Shape:", dot_product.shape)
print("Dot Product Matrix:\n", dot_product)
```
